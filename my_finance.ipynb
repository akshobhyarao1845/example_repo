```python
import csv

# 1. ADD INCOME / EXPENSE
def add_transaction():
    print("\n Add Income / Expense \n")
    type = input("Type (Income/Expense): ")
    try:
        amount = float(input("Amount: "))
        category = input("Category: ")

        with open("data.csv", "a", newline="") as f:
            writer = csv.writer(f)
            writer.writerow([type, amount, category])

        print("✔ Transaction added successfully!\n")
    except ValueError:
        print("Error: Please enter a number (like 100 or 100.5).")


# 2. SPLIT EXPENSE
def split_expense():
    print("\n--- Split Expense ---")
    total = float(input("Total bill amount: "))
    people = int(input("Number of friends: "))
    share = total / people
    print(f"Each person should pay ₹{share:.2f}\n")



# 3. LOAN EMI CALCULATOR

def loan_emi():
    print("\n--- Loan EMI Calculator ---")
    P = float(input("Loan amount: "))
    R = float(input("Annual interest rate (%): ")) / 12 / 100
    N = int(input("Months: "))

    EMI = (P * R * (1 + R)**N) / ((1 + R)**N - 1)
    print(f"Your monthly EMI is ₹{EMI:.2f}\n")



# 4. CURRENCY CONVERTER

def currency_converter():
    print("\n--- INR to USD Converter ---")
    rate = 0.012  # Example: 1 INR ≈ 0.012 USD
    inr = float(input("Enter amount in INR: "))
    usd = inr * rate
    print(f"₹{inr} = ${usd:.2f}\n")



# 5. SAVINGS GOAL TRACKER
def savings_goal():
    print("\n--- Savings Goal Tracker ---")
    goal = float(input("Your goal amount: "))
    saved = float(input("Amount saved so far: "))

    progress = (saved / goal) * 100
    print(f"You have reached {progress:.1f}% of your goal!\n")



# MAIN
def main():
    while True:
        print("\n===== MyFinance App =====")
        print("1. Add Income / Expense")
        print("2. Split Expense")
        print("3. Loan EMI")
        print("4. Currency Converter")
        print("5. Savings Goal")
        print("6. Exit")

        choice = input("Choose an option: ")

        if choice == "1":
            add_transaction()
        elif choice == "2":
            split_expense()
        elif choice == "3":
            loan_emi()
        elif choice == "4":
            currency_converter()
        elif choice == "5":
            savings_goal()
        elif choice == "6":
            print("Goodbye 👋")
            break
        else:
            print("❌ Invalid choice, try again.\n")


# Run the program
main()
```
